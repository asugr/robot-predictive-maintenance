# 🦾 Early-Warning Robot Failure Prediction

Can sensor measurements provide warning **before** failure begins?

## Target

For each motor run, identify the first failure onset. For each normal observation, predict whether failure begins within 1, 3, 5, or 10 seconds. Failure observations themselves are excluded.

## Validation

The final four operating conditions are held out completely to reduce time-series leakage.

```python
import pandas as pd
results=pd.read_csv('../data/processed/early_warning_results.csv')
display(results)
```

```python
import matplotlib.pyplot as plt
plt.plot(results.horizon_s,results.roc_auc,marker='o',label='ROC-AUC')
plt.plot(results.horizon_s,results.average_precision,marker='o',label='Average Precision')
plt.xlabel('Warning horizon (seconds)'); plt.ylabel('Score'); plt.legend(); plt.show()
```

## Engineering lesson

The practical question is not simply whether a model can recognize failure. It is whether it can provide enough warning to support an action. This creates a trade-off between **lead time** and **predictive quality**.
